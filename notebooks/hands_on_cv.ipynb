```python
# Solusi contoh untuk Latihan 3: to_binary_manual
def to_binary_manual(gray_img, thresh=128):
    """Ubah citra grayscale menjadi binary (0/255) tanpa cv2.threshold.
    gray_img : numpy array (H, W) dtype=uint8
    thresh : int (0-255)
    returns: binary image dtype=uint8 with values 0 or 255"""
    import numpy as _np
    if gray_img.dtype != _np.uint8:
        gray = gray_img.astype(_np.uint8)
    else:
        gray = gray_img
    # vektorisasi: perbandingan menghasilkan boolean, konversi ke uint8 dan kalikan 255
    binary = (gray > thresh).astype(_np.uint8) * 255
    return binary

# Demo: bandingkan dengan cv2.threshold
th = 128
binary_manual = to_binary_manual(gray, th)
ret, binary_cv = cv2.threshold(gray, th, 255, cv2.THRESH_BINARY)
print('cv2.threshold returned ret=', ret)
# tampilkan perbandingan
import matplotlib.pyplot as _plt
_plt.figure(figsize=(8,3))
_plt.subplot(1,3,1); _plt.title('Grayscale'); _plt.axis('off'); _plt.imshow(gray, cmap='gray')
_plt.subplot(1,3,2); _plt.title('Manual binary'); _plt.axis('off'); _plt.imshow(binary_manual, cmap='gray')
_plt.subplot(1,3,3); _plt.title('cv2.threshold'); _plt.axis('off'); _plt.imshow(binary_cv, cmap='gray')
_plt.tight_layout(); _plt.show()
# simpan hasil ke results/ untuk referensi
out_dir = Path('results')
out_dir.mkdir(exist_ok=True)
cv2.imwrite(str(out_dir / 'exercise3_binary_manual.png'), binary_manual)
cv2.imwrite(str(out_dir / 'exercise3_binary_cv.png'), binary_cv)
print('Saved binary images to', out_dir)
```
